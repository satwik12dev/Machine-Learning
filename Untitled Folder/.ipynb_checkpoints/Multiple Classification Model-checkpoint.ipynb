```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as mlp
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.metrics import accuracy_score, confusion_matrix

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
```

```python
sns.get_dataset_names()
```

```python
df = sns.load_dataset('titanic')
df
```

```python
df.info
```

```python
df.isnull().sum()
```

```python
df['age'] = df['age'].fillna(df['age'].median())
```

```python
df['embarked'] = df['embarked'].fillna("Unkown")
```

```python
df['embark_town'] = df['embark_town'].fillna("Not Available")
```

```python
features = ['pclass','sex','age','sibsp','parch','fare','embarked','class','alone']
```

```python
x = df[features]
y =df['survived']
```

```python
x
```

```python
y
```

```python
x = pd.get_dummies(x,columns=['sex','embarked','class','alone'],drop_first=True)
```

```python
x.head()
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
lr_model = LogisticRegression(max_iter=1000,random_state=42)
dt_model = DecisionTreeClassifier(max_depth=5,random_state=42)
rf_model = RandomForestClassifier(n_estimators=100,max_depth=5,random_state=42)
knn_model = KNeighborsClassifier(n_neighbors=5)
svn_model = SVC(kernel='rbf', probability=True, random_state=42)
nb_model = GaussianNB()
```

```python
lr_model.fit(x_train, y_train)
dt_model.fit(x_train, y_train)
rf_model.fit(x_train, y_train)
knn_model.fit(x_train, y_train)
svn_model.fit(x_train, y_train)
nb_model.fit(x_train, y_train)
```

```python
lr_y_pred  = lr_model.predict(x_test)
dt_y_pred  = dt_model.predict(x_test)
rf_y_pred  = rf_model.predict(x_test)
knn_y_pred  = knn_model.predict(x_test)
svn_y_pred  = svn_model.predict(x_test)
nb_y_pred  = nb_model.predict(x_test)
```

```python
lr_acc = accuracy_score(y_test, lr_y_pred)
dt_acc = accuracy_score(y_test, dt_y_pred)
rf_acc = accuracy_score(y_test, rf_y_pred)
knn_acc = accuracy_score(y_test, knn_y_pred)
svn_acc = accuracy_score(y_test, svn_y_pred)
nb_acc = accuracy_score(y_test, nb_y_pred)
```

```python
lr_acc
```

```python
result = [
    {'Model': 'Logistic Regression', 'Accuracy': lr_acc},
    {'Model': 'Decision Tree', 'Accuracy': dt_acc},
    {'Model': 'Random Forest', 'Accuracy': rf_acc},
    {'Model': 'KNN', 'Accuracy': knn_acc},
    {'Model': 'SVM', 'Accuracy': svn_acc},
    {'Model': 'Naive Bayes', 'Accuracy': nb_acc}
]
```

```python
result_df = pd.DataFrame(result)
```

```python
result_df
```

```python
mlp.barh(result_df['Model'],result_df['Accuracy'])
mlp.show()
```
